```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv(r'./Iris.csv')
df.info

y = df['Species']
X = df.drop(columns=['Id', 'Species'])
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=44, stratify=y)

X_train = np.asarray(X_train)
y_train = np.asarray(y_train)
X_test = np.asarray(X_test)
y_test = np.asarray(y_test)
```

```python
# def euclidean_distance(X_pred, x):
#     slDistance = x['SepalLengthCm'] - X_pred['SepalLength']
#     swDistance = x['SpealWidthCm'] - X_pred['SepalWidthCm']
#     plDistance = x['PetalLengthCm'] - X_pred['PetalLengthCm']
#     pwDistance = x['PetalWidthCm'] - X_pred['PetalWidthCm']
#     return np.sqrt(slDistance**2 + swDistance**2 + plDistance**2 + pwDistance**@)

def euclidean_distance(a,b):
    return np.sqrt(np.sum((np.asarray(a)-np.asarray(b))**2))
```

```python
def predict(val, k, X_train, y_train):
    distances = [euclidean_distance(val,x) for x in X_train]
    pairs = [(dist, label) for dist, label  in zip(distances, y_train)]
    pairs.sort()
    k_labels = [label for _, label in pairs[:k]]
    
    counts ={}
    for label in k_labels:
        counts[label] = counts.get(label, 0) +1
    pred = max(counts, key=counts.get)
    return pred

```

```python
from sklearn.metrics import classification_report, confusion_matrix
preds = [predict(x, 5, X_train, y_train) for x in X_test]
print(classification_report(y_test, preds))
```
